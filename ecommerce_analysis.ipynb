```python

# Import Libraries

import pandas as pd

# Load Datasets

customers = pd.read_csv(r"H:\DA projects\project_2\data_set\customers.csv")
orders = pd.read_csv(r"H:\DA projects\project_2\data_set\orders.csv")
order_items = pd.read_csv(r"H:\DA projects\project_2\data_set\order_items.csv")
products = pd.read_csv(r"H:\DA projects\project_2\data_set\products.csv")

```

```python
# Preview First 5 Rows of Each Dataset

print(customers.head())
print(orders.head())
print(order_items.head())
print(products.head())
```

```python
# produts Dataset Information
products["category"].unique()
```

```python
# Standardize Product Categories
products["category"] = (products["category"].str.strip().str.title())
```

```python
# Check Unique Product Categories

products["category"].unique()
```

```python
# Convert Date Columns
customers["signup_date"] = pd.to_datetime(customers["signup_date"])
orders["order_date"] = pd.to_datetime(orders["order_date"])
```

```python
# Verify Dataset Structure and Data Types
customers.info()
orders.info()
```

```python
# Check for Duplicate Records
print("Customers:", customers.duplicated().sum())
print("Orders:", orders.duplicated().sum())
print("Order Items:", order_items.duplicated().sum())
print("Products:", products.duplicated().sum())
```

```python
# Remove Duplicate Records
orders = orders.drop_duplicates()
```

```python
# Verify Duplicate Removal
orders.duplicated().sum()
```

```python

# Missing Value Analysis
print(customers.isnull().sum())
print()

print(orders.isnull().sum())
print()

print(order_items.isnull().sum())
print()

print(products.isnull().sum())
```

```python
# Inspect Missing Unit Price Records
order_items[order_items["unit_price"].isnull()].head()
```

```python
# Check Remaining Missing Unit Prices
order_items["unit_price"].isnull().sum()
```

```python
# Merge Product Price Information
order_items = order_items.merge(
    products[['product_id', 'list_price']],
    on='product_id',
    how='left'
)
```

```python
# Preview Order Items Dataset
order_items.head()
```

```python
# Impute Missing Unit Prices
order_items["unit_price"] = order_items["unit_price"].fillna(
    order_items["list_price"])
```

```python
# Verify Missing Unit Price Values
order_items["unit_price"].isnull().sum()
```

```python
# Remove Temporary Column
order_items.drop(
    columns=["list_price"],
    inplace=True
)
```

```python
# Merge Datasets
df = (
    orders
    .merge(customers, on="customer_id")
    .merge(order_items, on="order_id")
    .merge(products, on="product_id")
)
```

```python
# Check Shape of Merged Dataset
df.shape
```

```python
df.drop(
    columns=["list_price"],
    inplace=True
)
```
